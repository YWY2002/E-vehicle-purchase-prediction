```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("data/train-cat-encoded.csv")
print(df.columns)
print("Number of columns:", df.columns.size)
df.head()
```

### Correlation Heatmap

```python
corr = df.corr()
fig, ax = plt.subplots(figsize=(12, 10))
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(
    corr,
    mask=mask,
    annot=True,  # Print correlation coefficients
    fmt=".2f",  # 2 decimal places
    cmap="vlag",  # Diverging colormap centered at 0
    vmin=-1,
    vmax=1,  # Lock bounds to [-1, 1]
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8, "label": "Correlation Coefficient"},
)

ax.set_title("Training data Correlation Heatmap", fontsize=14, pad=12)
plt.tight_layout()
plt.show()
```
